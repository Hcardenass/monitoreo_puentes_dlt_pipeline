# Gold: Métricas en ventanas de 10 minutos que combinan agregados de silver

```python
import dlt
from pyspark.sql.functions import col, window, max, avg, round


@dlt.table(
    name="03_gold.puente_metrics",
    comment="10-min avg temperatura, max vibracion & max inclinación por puente con inicio/fin de ventana"
)
def puente_metrics():
    # Aplicar una marca de agua de 2 minutos para limitar los datos tardíos en operaciones con estado.
    temp = (
      dlt.read_stream("02_silver.puente_temperatura")
         .withWatermark("event_time", "2 minutes")
    )
    vib = (
      dlt.read_stream("02_silver.puente_vibracion")
         .withWatermark("event_time", "2 minutes")
    )
    tilt = (
      dlt.read_stream("02_silver.puente_inclinacion")
         .withWatermark("event_time", "2 minutes")
    )

    # Compute 10-minute tumbling average temperature, retaining metadata
    temp_agg = (
      temp
        .groupBy(
          window("event_time", "10 minutes"),
          col("puente_id"),
          col("name"),
          col("location")
        )
        .agg(
          avg("temperatura").alias("avg_temperatura")
        )
        .select(
          col("puente_id"),
          col("name"),
          col("location"),
          col("window.start").alias("window_start"),
          col("window.end").alias("window_end"),
          col("avg_temperatura")
        )
    )

    # Compute 10-minute max vibration per bridge
    vib_agg = (
      vib
        .groupBy(window("event_time", "10 minutes"), col("puente_id"))
        .agg(
          max("vibracion").alias("max_vibracion")
        )
        .select(
          col("puente_id"),
          col("window.start").alias("window_start"),
          col("window.end").alias("window_end"),
          col("max_vibracion")
        )
    )

    # Compute 10-minute max tilt angle per bridge
    tilt_agg = (
      tilt
        .groupBy(window("event_time", "10 minutes"), col("puente_id"))
        .agg(
          max("inclinacion").alias("max_inclinacion")
        )
        .select(
          col("puente_id"),
          col("window.start").alias("window_start"),
          col("window.end").alias("window_end"),
          col("max_inclinacion")
        )
    )

    # Join silver aggregates on bridge_id + window bounds
    return (
      temp_agg.alias("t")
        .join(
          vib_agg.alias("v"),
          on=["puente_id", "window_start", "window_end"],
          how="inner"
        )
        .join(
          tilt_agg.alias("l"),
          on=["puente_id", "window_start", "window_end"],
          how="inner"
        )
        .select(
          col("puente_id"),
          col("name"),
          col("location"),
          col("window_start"),
          col("window_end"),
          round(col("avg_temperatura"),2).alias("avg_temperatura"),
          col("max_vibracion"),
          col("max_inclinacion")
        )
    )
```
